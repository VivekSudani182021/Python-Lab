### 7. Write a program to calculate the area of a circle, triangle, and rectangle area using the math module. (B)

### 10. Write a program to print the person's age in years and also print how many days remain for the next birthday. (B) 

```python
from datetime import date

dob_year = int(input("Enter birth year: "))
dob_month = int(input("Enter birth month: "))
dob_day = int(input("Enter birth day: "))

today = date.today()
dob = date(dob_year, dob_month, dob_day)

# Age in years
age = today.year - dob.year
if (today.month, today.day) < (dob.month, dob_day):
    age -= 1

# Next birthday
next_bday_year = today.year
if (today.month, today.day) >= (dob_month, dob_day):
    next_bday_year += 1

next_bday = date(next_bday_year, dob_month, dob_day)
days_left = (next_bday - today).days

print("Age:", age)
print("Days remaining for next birthday:", days_left)
```
