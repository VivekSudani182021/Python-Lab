### 01) Create a dictionary for N values & print the size (A)

```python
N = int(input("Enter number of elements: "))
d = {}

for i in range(N):
    key = input("Enter key: ")
    value = input("Enter value: ")
    d[key] = value

print("Dictionary:", d)
print("Size of dictionary:", len(d))
```

### 02) Create a dictionary from a string. (A)

```python
s = input("Enter a string: ")
d = {}

for ch in s:
    d[ch] = d.get(ch, 0) + 1

print("Dictionary from string:", d)
```

### 03) Sort a dictionary by key (ascending & descending).

```python
d = {'c': 3, 'a': 1, 'b': 2}

asc = dict(sorted(d.items()))
desc = dict(sorted(d.items(), reverse=True))

print("Ascending:", asc)
print("Descending:", desc)
```

### 04) Add a key to dictionary if it does not exist. (A)

```python
d = {'a': 1, 'b': 2}

key = input("Enter key to add: ")
value = input("Enter value: ")

if key not in d:
    d[key] = value
    print("Key added.")
else:
    print("Key already exists.")

print("Updated dictionary:", d)
```

### 05) Sort dictionary by value (ascending & descending). (B)

```python
d = {'a': 3, 'b': 1, 'c': 2}

asc = dict(sorted(d.items(), key=lambda item: item[1]))
desc = dict(sorted(d.items(), key=lambda item: item[1], reverse=True))

print("Ascending by value:", asc)
print("Descending by value:", desc)
```

### 06) Remove a key from dictionary if it exists. (B)

```python
d = {'a': 1, 'b': 2, 'c': 3}

key = input("Enter key to remove: ")

if key in d:
    d.pop(key)
    print("Key removed.")
else:
    print("Key not found.")

print("Updated dictionary:", d)
```

### 07) Merge two dictionaries. (B)

```python
d1 = {}
d2 = {}

n1 = int(input("Enter number of elements in dictionary 1: "))
for i in range(n1):
    k = input("Key: ")
    v = input("Value: ")
    d1[k] = v

n2 = int(input("Enter number of elements in dictionary 2: "))
for i in range(n2):
    k = input("Key: ")
    v = input("Value: ")
    d2[k] = v

merged = {**d1, **d2}

print("Merged dictionary:", merged)
```

### 8)Convert two lists into a dictionary. (B)

```python
keys = ['a', 'b', 'c']
values = [1, 2, 3]

d = dict(zip(keys, values))

print("Dictionary:", d)
```
