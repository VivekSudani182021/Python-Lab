### 1. Write a program to calculate simple interest using a function. (A) 

```python
def simple_interest(p, r, t):
    return (p * r * t) / 100

p = float(input("Enter Principal: "))
r = float(input("Enter Rate: "))
t = float(input("Enter Time: "))

print("Simple Interest =", simple_interest(p, r, t))
```

### 2. Write a program that defines a function to add first N numbers. (A)


```python
def sum_n(n):
    s = 0
    for i in range(1, n+1):
        s += i
    return s

n = int(input("Enter N: "))
print("Sum of first N numbers =", sum_n(n))
```

### 3. Write a program to find the maximum number from the two given numbers using a function. (A)


```python
def maximum(a, b):
    if a > b:
        return a
    else:
        return b

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("Maximum number is =", maximum(a, b))
```

### 4. Write a program to take a string from the user & pass it as an argument and convert all lowercase characters into uppercase using a function. (A)

```python
def to_upper(s):
    return s.upper()

s = input("Enter a string: ")
print("Uppercase String =", to_upper(s))
```

### 5. Write a program to find the factorial of a given number using function. (A) 

```python
def factorial(n):
    f = 1
    for i in range(1, n+1):
        f *= i
    return f

n = int(input("Enter a number: "))
print("Factorial =", factorial(n))
```

### 6. Write a program to generate a Fibonacci series using a function. (e.g. 0 1 1 2 3 5 8..). (A)

```python
def fibonacci(n):
    a, b = 0, 1
    for i in range(n):
        print(a, end=" ")
        a, b = b, a + b

n = int(input("Enter number of terms: "))
fibonacci(n)
```

### 7. Write a program to implement a simple calculator using the lambda function. (A)

```python
add = lambda a, b: a + b
sub = lambda a, b: a - b
mul = lambda a, b: a * b
div = lambda a, b: a / b

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Addition =", add(a, b))
print("Subtraction =", sub(a, b))
print("Multiplication =", mul(a, b))
print("Division =", div(a, b))
```

### 8. Write a program that defines a function that returns 1 if the number is prime otherwise, return 0. (B)

```python
def is_prime(n):
    if n <= 1:
        return 0
    for i in range(2, n):
        if n % i == 0:
            return 0
    return 1

n = int(input("Enter a number: "))
print(is_prime(n))
```

### 9. Write a program to find the factorial of a given number using recursion. (B)

```python
def fact(n):
    if n==0 or n==1:
        return 1
    else:
        return n * fact(n-1)
fact(5)     
```

### 10. Write a program to generate a Fibonacci series of a given number N using recursion.(B)

```python
def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)

N = int(input("Enter number of terms: "))

for i in range(N):
    print(fib(i), end=" ")
```
