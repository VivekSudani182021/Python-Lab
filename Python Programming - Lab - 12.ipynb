### 1. Write a program to read a file named firstfile.txt. (A) 

```python
f = open("firstfile.txt", "r")
content = f.read()
print(content)
f.close()
```

### 2. Write a program to read the first 5 lines from the file named firstfile.txt. (A) 


```python
f = open("firstfile.txt", "r")
for i in range(5):
    line = f.readline()
    if not line:
        break
    print(line, end="")
f.close()
```


### 3. Write a program to read only special characters from a file. (A) 

```python
f = open("firstfile.txt", "r")
data = f.read()

for ch in data:
    if not ch.isalnum() and not ch.isspace():
        print(ch, end=" ")

f.close()
```

### 4. Write a program to read file line by line and store lines as a list. (A)

### 5. Write a program to write N lines in a new file. (A) 


```python
n = int(input("Enter number of lines: "))
f = open("newfile.txt", "w")

for i in range(n):
    line = input(f"Enter line {i+1}: ")
    f.write(line + "\n")

f.close()
print("File written successfully.")
```

### 6. Write a program to write 5 student records (Roll No, Student Name, and Department) in the studentDetails.txt file. (A) 

```python
f = open("studentDetails.txt", "w")

for i in range(5):
    roll = input("Enter Roll No: ")
    name = input("Enter Name: ")
    dept = input("Enter Department: ")
    f.write(roll + ", " + name + ", " + dept + "\n")

f.close()
print("5 records saved.")
```

### 7. Write a program to find the longest word in a file named firstfile.txt. (B) 


```python
f = open("firstfile.txt", "r")
words = f.read().split()
f.close()

longest = max(words, key=len)
print("Longest word is:", longest)
```

### 9. Write a program to append the content of the studentDetails.txt file by reading the student records from the user. (B)

```python
f = open("studentDetails.txt", "a")

n = int(input("How many student records to append? "))

for i in range(n):
    roll = input("Enter Roll No: ")
    name = input("Enter Name: ")
    dept = input("Enter Department: ")
    f.write(roll + ", " + name + ", " + dept + "\n")

f.close()
print("Records appended successfully.")
```

### 10. Write a program to enter a file name and check whether it exists or not. Ask for user confirmation and delete the file. (B) 

```python
import os

filename = input("Enter file name: ")

if os.path.exists(filename):
    print("File exists.")
    choice = input("Do you want to delete it? (yes/no): ")

    if choice.lower() == "yes":
        os.remove(filename)
        print("File deleted.")
    else:
        print("File not deleted.")
else:
    print("File does not exist.")

```
