### Problem 1: Create a `Student` class

```python
class Student:
    def __init__(self, enrollment_no, student_name, gender, department):
        self.enrollment_no = enrollment_no
        self.student_name = student_name
        self.gender = gender
        self.department = department

    def display_student_info(self):
        print(f"Enrollment No: {self.enrollment_no}")
        print(f"Student Name: {self.student_name}")
        print(f"Gender: {self.gender}")
        print(f"Department: {self.department}")

student1 = Student("23010101260", "Vivek Sudani", "Male", "Computer Science")

print("Student 1 Information:")
student1.display_student_info()
```

### Problem 2: `Shape` class and its subclasses (`Circle`, `Triangle`, `Square`)

```python
import math

class Shape:
    def CalculateArea(self):
        raise NotImplementedError("Subclasses must implement this method")

    def DisplayArea(self):
        raise NotImplementedError("Subclasses must implement this method")

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
        self.area = 0

    def CalculateArea(self):
        self.area = math.pi * (self.radius ** 2)

    def DisplayArea(self):
        print(f"Area of Circle (radius {self.radius}): {self.area:.2f}")

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height
        self.area = 0

    def CalculateArea(self):
        self.area = 0.5 * self.base * self.height

    def DisplayArea(self):
        print(f"Area of Triangle (base {self.base}, height {self.height}): {self.area:.2f}")

class Square(Shape):
    def __init__(self, side):
        self.side = side
        self.area = 0

    def CalculateArea(self):
        self.area = self.side ** 2

    def DisplayArea(self):
        print(f"Area of Square (side {self.side}): {self.area:.2f}")

print("\n--- Shape Calculations ---")

circle = Circle(5)
circle.CalculateArea()
circle.DisplayArea()

triangle = Triangle(10, 4)
triangle.CalculateArea()
triangle.DisplayArea()

square = Square(7)
square.CalculateArea()
square.DisplayArea()
```

### Problem 3: Create a `Bank_Account` class

```python
class Bank_Account:
    def __init__(self, account_no, user_name, email, account_type, account_balance):
        self.account_no = account_no
        self.user_name = user_name
        self.email = email
        self.account_type = account_type
        self.account_balance = account_balance

    def GetAccountDetails(self, account_no, user_name, email, account_type, account_balance):

        self.account_no = account_no
        self.user_name = user_name
        self.email = email
        self.account_type = account_type
        self.account_balance = account_balance
        print("Account details updated/set.")

    def DisplayAccountDetails(self):
        print(f"\nAccount Number: {self.account_no}")
        print(f"User Name: {self.user_name}")
        print(f"Email: {self.email}")
        print(f"Account Type: {self.account_type}")
        print(f"Account Balance: ${self.account_balance:.2f}")

if __name__ == "__main__":
    print("\n--- Bank Account Demonstration ---")

    my_account = Bank_Account("BA0012345", "John Doe", "john.doe@example.com", "Savings", 1500.75)
    my_account.DisplayAccountDetails()

    my_account.GetAccountDetails("BA0012345", "John A. Doe", "john.a.doe@example.com", "Checking", 2500.50)
    my_account.DisplayAccountDetails()

    jane_account = Bank_Account("BA0067890", "Jane Smith", "jane.smith@example.com", "Current", 5000.00)
    jane_account.DisplayAccountDetails()
```
