### Program 1: Basic Calculator with Exception Handling

```python
def calculator():
    try:
        num1_str = input("Enter the first number: ")
        num2_str = input("Enter the second number: ")
        operator = input("Enter the operator (+, -, *, /): ")

        num1 = float(num1_str)
        num2 = float(num2_str)

        if operator == '+':
            result = num1 + num2
        elif operator == '-':
            result = num1 - num2
        elif operator == '*':
            result = num1 * num2
        elif operator == '/':
            if num2 == 0:
                raise ZeroDivisionError("Cannot divide by zero!")
            result = num1 / num2
        else:
            raise ValueError("Invalid operator. Please use +, -, *, or /.")

        print(f"Result: {num1} {operator} {num2} = {result}")

    except ValueError as e:
        print(f"Input Error: {e}. Please ensure you enter valid numbers and a valid operator.")
    except ZeroDivisionError as e:
        print(f"Calculation Error: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

calculator()
```

### Program 2: Blood Donation Eligibility Check with Custom Exception

```python
class BloodDonationCriteriaError(Exception):
    """Custom exception for blood donation criteria not met."""
    pass

def check_blood_donation_eligibility():
    try:
        age_str = input("Enter your age: ")
        weight_str = input("Enter your weight in kg: ")

        age = int(age_str)
        weight = float(weight_str)

        if age <= 0 or weight <= 0:
            raise ValueError("Age and weight must be positive values.")

        if age > 18 and weight > 50:
            print("Congratulations! You are eligible to donate blood.")
        else:
            reasons = []
            if age <= 18:
                reasons.append("Age must be greater than 18 years.")
            if weight <= 50:
                reasons.append("Weight must be greater than 50 kg.")
            raise BloodDonationCriteriaError("\n".join(reasons))

    except ValueError as e:
        print(f"Invalid Input Error: {e}. Please enter valid numbers for age and weight.")
    except BloodDonationCriteriaError as e:
        print(f"Eligibility Criteria Not Met:\n{e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

check_blood_donation_eligibility()
```

### Program 3: Division with Custom Exception for Zero Division

```python
class DivideByZeroCustomError(Exception):
    """Custom exception for division by zero."""
    pass

def custom_division():
    try:
        num1_str = input("Enter the dividend: ")
        num2_str = input("Enter the divisor: ")

        num1 = float(num1_str)
        num2 = float(num2_str)

        if num2 == 0:
            raise DivideByZeroCustomError("Cannot perform division by zero using the custom exception.")

        result = num1 / num2
        print(f"Result of division: {num1} / {num2} = {result}")

    except ValueError:
        print("Input Error: Please ensure you enter valid numbers.")
    except DivideByZeroCustomError as e:
        print(f"Calculation Error: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

custom_division()
```
