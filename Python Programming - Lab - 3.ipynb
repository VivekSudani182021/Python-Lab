### 12)Celsius ↔ Fahrenheit converter.

```python
c = float(input("Enter temperature in Celsius: "))
f = (9 * c / 5) + 32
print("Fahrenheit =", f)

```

```python
f2 = float(input("Enter temperature in Fahrenheit: "))
c2 = (f2 - 32) * 5 / 9
print("Celsius =", c2)
```
