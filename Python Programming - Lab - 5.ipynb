### 01) Write a program to find the length of a string with and without using the len() 

```python
### 1)
string = input("Enter a string: ")
count = 0

for ch in string:
    count += 1

print("Length of string (without len):", count)

### 2)
string = input("Enter a string: ")
print("Length of string (using len):", len(string))
```

### 02)Write a program to remove ith character from the given string. (A)

```python
string = input("Enter a string: ")
i = int(input("Enter the index to remove: "))   

if 0 <= i < len(string):
    new_string = string[:i] + string[i+1:]
    print("String after removing character:", new_string)
else:
    print("Invalid index!")
```

### 04) Write a program to count the number of vowels in a given string. (A)

```python
s = input("Enter a string: ")

vowels = "aeiouAEIOU"
count = 0

for ch in s:
    if ch in vowels:
        count += 1

print("Total vowels:", count)

```

### 05) 5. Write a program to print even-length words in a string. (A)  


```python
s = input("Enter a string: ")

words = s.split()   

print("Even-length words:")
for w in words:
    if len(w) % 2 == 0:
        print(w)

```

### 06)6. Write a program to check whether the given string is a palindrome or not. (B)  


```python
s = input("Enter a string: ")

if s == s[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

### 07) Write a program to do string slicing from left rotate and right rotate the given string 
### 
by d elements (where d <= n). (B)  

```python
s = input("Enter a string: ")
d = int(input("Enter d (number of rotations): "))

n = len(s)
d = d % n          

left_rotate = s[d:] + s[:d]


right_rotate = s[n-d:] + s[:n-d]

print("Left Rotation :", left_rotate)
print("Right Rotation:", right_rotate)
```

### 08) Write a program to reverse the words in a given string. (B) 0

```python
s = input("Enter a string: ")

words = s.split()           
rev_words = words[::-1]     

result = " ".join(rev_words)

print("Reversed words:", result)


```
