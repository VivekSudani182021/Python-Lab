### 01) Print 1 to N using for & while loop

```python
N = int(input("Enter N: "))
for i in range(1, N+1):
    print(i, end=" ")

```

```python
N = int(input("Enter N: "))
i = 1
while i <= N:
    print(i, end=" ")
    i += 1
```

### 2. Sum of N numbers using for & while loop

```python
N = int(input("Enter N: "))
total = 0
for i in range(1, N+1):
    total += i
print("Sum =", total)
```

```python
N = int(input("Enter N: "))
i = 1
total = 0
while i <= N:
    total += i
    i += 1
print("Sum =", total)
```

### 03)Factorial of a number

```python
n = int(input("Enter number: "))
fact = 1

for i in range(1, n+1):
    fact *= i

print("Factorial =", fact)
```

### 05) WAP to check whether the given year is leap year or not.
[If a year can be divisible by 4 but not divisible by 100 then it is leap year but if it is divisible by 400 then it is leap year]

```python
year=int(input("Enter Year"))
if(year%4==0 and (year%100!=0 or year%400==0)):
    print("Leap year")
else:
    print("not Leap year")
    
```

### 07) WAP to implement simple calculator which performs (add,sub,mul,div) of two no. based on user input.

```python
n1=int(input("Enter First Number : "))
n2=int(input("Enter Second Number : "))
ch=input("Enter Operator")
if(ch=='+'):print(str(n1)+" + "+str(n2)+" = ",(n1+n2))
elif(ch=='-'):print(str(n1)+" - "+str(n2)+" = ",(n1-n2))
elif(ch=='*'):print(str(n1)+" * "+str(n2)+" = ",(n1*n2))
elif(ch=='/'):print(str(n1)+" / "+str(n2)+" = ",(n1/n2))
else:print("Enter valid operator")
```

### 8) WAP to calculate electricity bill based on following criteria. Which takes the unit from the user.
a. First 1 to 50 units – Rs. 2.60/unit</br>
b. Next 50 to 100 units – Rs. 3.25/unit</br>
c. Next 100 to 200 units – Rs. 5.26/unit</br>
d. above 200 units – Rs. 8.45/unit

```python
units=int(input("Enter Units"))
bill=0
if(units<=50):
    bill=units*2.60
elif(units>50 and units<=100):
    bill+=(units-50)*3.25+2.60*50
elif(units>100 and units<=200):
    bill+=(units-100)*5.26+50*3.60+50*2.60
elif(units>200 ):
    bill+=(units-200)*8.45+50*3.60+50*2.60+100*5.26

print("Total Bill : ",bill)

```
