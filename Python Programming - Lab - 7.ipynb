```python
n = int(input("Enter the Number: "))

# Upper part of the diamond
for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")
    for k in range(i):
        print("* ", end="")
    print()

# Lower part of the diamond
for i in range(n - 1, 0, -1):
    for j in range(n - i):
        print(" ", end="")
    for k in range(i):
        print("* ", end="")
    print()
```

```python
rows = int(input("Enter number of rows: "))

pascal = [[1]]

for i in range(1, rows):
    row = [1]
    for j in range(1, i):
        row.append(pascal[i-1][j-1] + pascal[i-1][j])
    row.append(1)
    pascal.append(row)

for r in pascal:
    print(*r)

```

```python
n = int(input("Enter the Number: "))

# Upper part of the diamond
for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")
    for k in range(i):
        print(f"{i} " , end="")
    print()

# Lower part of the diamond
for i in range(n - 1, 0, -1):
    for j in range(n - i):
        print(" ", end="")
    for k in range(i):
        print(f"{i} ", end="")
    print()
```
