### 01) Sum of all elements in the list (A)

```python
lst = list(map(int, input("Enter numbers: ").split()))
print("Sum =", sum(lst))
```

### 02) Smallest & Largest element from list (A)

```python
lst = list(map(int, input("Enter numbers: ").split()))
print("Smallest =", min(lst))
print("Largest =", max(lst))
```

### 03) Split list into two parts & append first part to end

```python
lst = input("Enter list elements: ").split()
pos = int(input("Enter split position: "))
first = lst[:pos]
second = lst[pos:]
new_list = second + first
print("New List:", new_list)
```

### 04)Interchange first & last element (A).

```python
lst = input("Enter list elements: ").split()
lst[0], lst[-1] = lst[-1], lst[0]
print("After swapping:", lst)
```

### 05)Reverse the list (A)


```python
lst = input("Enter list elements: ").split()
print("Reversed List:", lst[::-1])
```

### 06) Print all even numbers from list (A)

```python
lst = list(map(int, input("Enter numbers: ").split()))
even = [x for x in lst if x % 2 == 0]
print("Even numbers:", even)
```

### 07)Search for an element in list (A)

```python
lst = input("Enter list elements: ").split()
item = input("Enter element to search: ")

if item in lst:
    print("Element found")
else:
    print("Element not found")
```

### 8) Sort list in ascending order (A)

```python
### 12) Product details with total amount (C)
```

```python
products = []
FinalAmount=0
productsCount = int(input("Enter How Many prodocuts you want to Enter"))
for i in range(productsCount):
    name = input(f"Enter product {i+1} name: ")
    qty = int(input("Enter quantity: "))
    price = float(input("Enter price: "))
    total = qty * price
    products.append([name, qty, price, total])
    print()

print("\nProduct Details:")
print("Name\tQty\tPrice\tTotal")
for p in products:
    print(f"{p[0]}\t{p[1]}\t{p[2]}\t{p[3]}")
    FinalAmount +=p[3]
print("Final amount : ",FinalAmount)
```
