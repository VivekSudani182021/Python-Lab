### 01) Create a set using list & find its size (A)

```python
lst = input("Enter elements: ").split()
st = set(lst)
print("Set:", st)
print("Size of set:", len(st))
```

### 02)Find maximum & minimum from set (A)

```python
a = set()
size = int(input("Enter the Size of set: "))
for i in range(size):
    n = int(input("Enter an element in set: "))
    a.add(n)
print(a)
```

### 03)Remove element from set (A)

```python
st = set(input("Enter elements: ").split())
item = input("Enter element to remove: ")

st.discard(item)   
print("Updated set:", st)
```

### 04) Convert set → tuple AND tuple → set (B)

```python
st = set(input("Enter set elements: ").split())
tp = tuple(st)

print("Tuple:", tp)

tp2 = tuple(input("Enter tuple elements: ").split())
st2 = set(tp2)

print("Set:", st2)
```

### 05) Union, Intersection, Difference, Symmetric Difference (B)

```python
set1 = set(input("Enter first set elements: ").split())
set2 = set(input("Enter second set elements: ").split())

print("Union:", set1 | set2)
print("Intersection:", set1 & set2)
print("Difference:", set1 - set2)
print("Symmetric Difference:", set1 ^ set2)
```

### 06) Enter values & reverse tuple (A)

```python
tp = tuple(input("Enter elements: ").split())
rev = tp[::-1]
print("Reversed tuple:", rev)
```

### 07) Remove duplicate values from tuple (A)

```python
tp = tuple(input("Enter elements: ").split())
unique = tuple(set(tp))
print("After removing duplicates:", unique)
```

### 8) Check if tuple is distinct (A)

```python
tp = tuple(input("Enter elements: ").split())

if len(tp) == len(set(tp)):
    print("Tuple is distinct")
else:
    print("Tuple is NOT distinct")
```

```python
### 9) Find tuples with positive elements (A)
```

```python
lst = eval(input("Enter list of tuples: "))

positive = [t for t in lst if all(x > 0 for x in t)]
print("Tuples with all positive elements:", positive)
```

```python
### 10) Tuples where all elements are divisible by k (B)
```

```python
lst = eval(input("Enter list of tuples: "))
k = int(input("Enter value of k: "))

result = [t for t in lst if all(x % k == 0 for x in t)]
print("Tuples divisible by", k, ":", result)
```
